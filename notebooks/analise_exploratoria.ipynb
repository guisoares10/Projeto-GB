## 1.1 `qt_material` e itens por pedido

`qt_material` (itens vendidos) é o **target** do forecast. Em vez de repetir todas as visões para `nr_pedidos`, usamos o apoio na forma de **itens por pedido** (`qt_material / nr_pedidos`, tamanho médio da cesta): cada visão mostra o volume de itens e, logo abaixo, quantos itens vão em cada pedido. Roteiro: geral → linha do tempo → canal → categoria → App × Site por categoria.

```python
metricas_volume = ["qt_material", "itens_por_pedido"]

kpi.geral(metricas_volume)
kpi.linha_do_tempo(metricas_volume)
kpi.por_canal(metricas_volume)
kpi.por_categoria(metricas_volume)
kpi.canal_x_categoria(metricas_volume)
```

```python
# App x Site têm o mesmo mix? Pareia o share de cada (mês, categoria) nos dois canais
mix_mensal = agregar(
    base_kpi[base_kpi["des_categoria_material"] != "CATEGORIA_INVALIDA"],
    ["mes", "des_canal_venda_final_agrup", "des_categoria_material"],
)
mix_mensal["share_qt_material_mes_canal"] = (
    mix_mensal["qt_material"]
    / mix_mensal.groupby(["mes", "des_canal_venda_final_agrup"])["qt_material"].transform("sum")
)
mix_canal = (
    mix_mensal
    .pivot_table(index=["mes", "des_categoria_material"], columns="des_canal_venda_final_agrup",
                 values="share_qt_material_mes_canal")
    .dropna()
    * 100
)

# Nível: mesma composição? | Movimento: removida a média de cada categoria, as variações andam juntas?
corr_nivel = mix_canal["App"].corr(mix_canal["Site"])
desvio = mix_canal - mix_canal.groupby(level="des_categoria_material").transform("mean")
corr_movimento = desvio["App"].corr(desvio["Site"])

display(pd.DataFrame({
    "medida": [
        "Correlação dos shares (nível)",
        "Correlação das variações mensais (movimento)",
        "Diferença média absoluta (p.p.)",
        "Pares mês × categoria comparados",
    ],
    "valor": [
        round(corr_nivel, 3),
        round(corr_movimento, 3),
        round((mix_canal["App"] - mix_canal["Site"]).abs().mean(), 2),
        len(mix_canal),
    ],
}))
```

### Quantidade × receita: o volume extra vem mais barato?

O preço por item é a razão `receita_aprovada / qt_material`. Acima ele foi visto por mês, canal e categoria; aqui olhamos a **relação dia a dia** entre as duas pontas da razão — a receita acompanha o volume na mesma proporção?

- **Dispersão diária:** cada ponto é um dia, colorido pela taxa de desconto. A linha tracejada é a receita **proporcional** ao volume (preço por item fixo no preço médio do período); a linha cheia é o ajuste real.
- **Elasticidade:** inclinação do ajuste em log (log receita = a + b · log itens). b = 1 significa receita proporcional ao volume; b < 1 significa que cada item a mais traz menos receita, ou seja, o preço por item cai quando o volume sobe.
- **Faixas de volume:** os dias são divididos em 5 grupos de mesmo tamanho (quintis de `qt_material`), comparando preço por item e taxa de desconto.

Usamos a série diária (`daily`), com a receita como veio na base (a correção de sinal da seção 2.4 muda o nível, não a relação).

```python
dia = daily[["data", "qt_material", "receita_aprovada", "vlr_venda_desconto"]].copy()
for coluna in ["qt_material", "receita_aprovada", "vlr_venda_desconto"]:
    dia[coluna] = dia[coluna].astype(float)
dia["preco_medio_item"] = dia["receita_aprovada"] / dia["qt_material"]
dia["taxa_desconto"] = dia["vlr_venda_desconto"] / (dia["receita_aprovada"] + dia["vlr_venda_desconto"])
dia["novembro"] = pd.to_datetime(dia["data"]).dt.month == 11

# Elasticidade: inclinação de log(receita) contra log(itens)
b_todos, a_todos = np.polyfit(np.log(dia["qt_material"]), np.log(dia["receita_aprovada"]), 1)
fora_nov = dia[~dia["novembro"]]
b_fora, _ = np.polyfit(np.log(fora_nov["qt_material"]), np.log(fora_nov["receita_aprovada"]), 1)
top_qt = set(dia.nlargest(20, "qt_material")["data"])
top_receita = set(dia.nlargest(20, "receita_aprovada")["data"])
display(pd.DataFrame({
    "medida": [
        "Correlação diária itens × receita",
        "Correlação diária itens × preço por item",
        "Elasticidade (todos os dias)",
        "Elasticidade (sem novembro)",
        "Dobrar os itens multiplica a receita por",
        "Dias em comum entre os 20 de maior volume e os 20 de maior receita",
    ],
    "valor": [
        round(dia["qt_material"].corr(dia["receita_aprovada"]), 3),
        round(dia["qt_material"].corr(dia["preco_medio_item"]), 3),
        round(b_todos, 2),
        round(b_fora, 2),
        round(2 ** b_todos, 2),
        len(top_qt & top_receita),
    ],
}))

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(17, 5), gridspec_kw={"width_ratios": [1.6, 1, 1]})
pontos = ax1.scatter(dia["qt_material"] / 1e3, dia["receita_aprovada"] / 1e6, c=dia["taxa_desconto"],
                     cmap="Blues", vmin=0.15, vmax=0.6, s=28, edgecolor="#1f4e79", linewidth=0.4)
ax1.scatter(dia.loc[dia["novembro"], "qt_material"] / 1e3, dia.loc[dia["novembro"], "receita_aprovada"] / 1e6,
            facecolor="none", edgecolor="#d17a22", s=60, linewidth=1.2, label="dias de novembro")
x = np.linspace(dia["qt_material"].min(), dia["qt_material"].max(), 200)
preco_periodo = dia["receita_aprovada"].sum() / dia["qt_material"].sum()
ax1.plot(x / 1e3, x * preco_periodo / 1e6, color="#7a7a7a", ls="--", lw=1.2,
         label=f"proporcional (R$ {preco_periodo:.2f} por item)")
ax1.plot(x / 1e3, np.exp(a_todos) * x ** b_todos / 1e6, color="#1f4e79", lw=2,
         label=f"ajuste real (elasticidade {b_todos:.2f})")
ax1.set(xlabel="itens vendidos no dia (mil)", ylabel="receita do dia (R$ mi)",
        title="Receita × quantidade por dia")
ax1.legend(loc="upper left", fontsize=9)
ax1.grid(alpha=0.3)
plt.colorbar(pontos, ax=ax1, label="taxa de desconto do dia", format=lambda v, _: f"{v:.0%}")

dia["faixa_volume"] = pd.qcut(dia["qt_material"], 5, labels=["1 (menor)", "2", "3", "4", "5 (maior)"])
faixas = dia.groupby("faixa_volume", observed=True).agg(
    dias=("data", "size"),
    itens_mediana=("qt_material", "median"),
    preco_medio_item=("preco_medio_item", "median"),
    taxa_desconto=("taxa_desconto", "median"),
    dias_novembro=("novembro", "sum"),
)
ax2.bar(faixas.index.astype(str), faixas["preco_medio_item"], color="#1f4e79")
ax2.set(title="Preço por item (mediana)", xlabel="faixa de volume do dia", ylabel="R$ por item")
ax3.bar(faixas.index.astype(str), faixas["taxa_desconto"], color="#a23b72")
ax3.set(title="Taxa de desconto (mediana)", xlabel="faixa de volume do dia")
ax3.yaxis.set_major_formatter(lambda v, _: f"{v:.0%}")
for ax, coluna, formato in [(ax2, "preco_medio_item", "R$ {:.0f}"), (ax3, "taxa_desconto", "{:.0%}")]:
    for posicao, valor in enumerate(faixas[coluna]):
        ax.text(posicao, valor, formato.format(valor), ha="center", va="bottom", fontsize=9)
    ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
display(faixas)
```

#### Insights — quantidade × receita

- **Receita e volume andam juntos** (correlação diária de **0,95**; 18 dos 20 dias de maior volume também estão entre os 20 de maior receita). Prever bem `qt_material` já dá uma boa leitura da receita.
- **Mas a receita cresce menos que o volume:** a elasticidade é **0,71** — **dobrar os itens vendidos multiplica a receita por ~1,6**, não por 2. Sem novembro, a elasticidade fica em 0,75: não é só efeito da Black Friday.
- **O volume extra vem mais barato:** dos dias de menor volume para os de maior, o preço por item cai de **R$ 66 para R$ 42** e a taxa de desconto sobe de **29% para 51%**. Os dias de maior volume são, em grande parte, dias de desconto alto (28 dos 49 dias da faixa mais alta são de novembro).
- **Implicação para o negócio:** um pico de itens não significa um pico de receita do mesmo tamanho. Ao planejar uma campanha, o ganho de volume precisa ser comparado à receita sacrificada no desconto — e, para o modelo, indica que **o desconto é a variável que mais acompanha o volume** (seção 1.4). É uma associação, não prova de causa: no modelo, o desconto aparece como a variável temporal e acionável de maior contribuição preditiva.

## 1.4 Desconto

Primeiro confirmamos como receita e desconto se relacionam; depois seguimos o mesmo roteiro da 1.1 com a **taxa de desconto** = `vlr_venda_desconto / (receita_aprovada + vlr_venda_desconto)`.

### A `receita_aprovada` já vem com o desconto aplicado?

A descrição do case diz apenas "receita gerada na venda". Antes de analisar desconto, precisamos saber como as duas colunas se relacionam:

- **Hipótese A (receita líquida):** `receita_aprovada` é o que o cliente pagou, já com desconto. Preço cheio = `receita_aprovada + vlr_venda_desconto`.
- **Hipótese B (receita bruta):** `receita_aprovada` é o preço cheio. Valor pago = `receita_aprovada - vlr_venda_desconto`.

Foi testado as duas com três verificações. Só considerado linhas com receita ≥ 0.

```python
sql_revenue_semantics = carregar_sql("04_sql_revenue_semantics.sql")

revenue_semantics = run_bq(sql_revenue_semantics, "semântica receita x desconto")
display(revenue_semantics.T.rename(columns={0: "valor"}))

sql_price_hypotheses = carregar_sql("05_sql_price_hypotheses.sql")

price_hypotheses = run_bq(sql_price_hypotheses, "preço por item em cada hipótese")
display(price_hypotheses)

sql_free_items = carregar_sql("06_sql_free_items.sql")

print("Exemplos de linhas com receita zero e desconto positivo:")
display(run_bq(sql_free_items, "itens com receita zero"))
```

**Conclusão: `receita_aprovada` é receita líquida (hipótese A).**

1. **Na hipótese B haveria vendas negativas.** Em ~13% das linhas o desconto é maior que a receita; o valor pago seria negativo, e o preço médio pago por item em FACIAL ficaria abaixo de zero (e PERFUMARIA FEMININA a ~R$ 4). Na hipótese A os preços cheios são plausíveis para o catálogo (perfumaria masculina ~R$ 138, feminina ~R$ 119).
2. **Linhas com receita zero são itens gratuitos.** Elas têm desconto positivo, e o desconto por item tem cara de preço de catálogo (R$ 39,90 em FACIAL, R$ 64,90 em CORPO E BANHO): o item saiu 100% de graça (brinde). Na hipótese B seria um produto de preço zero com desconto, o que é impossível.
3. **A taxa só faz sentido na hipótese A.** Em A a taxa por linha tem mediana de ~31% e nunca passa de 100% (chega a 100% exatamente nos itens gratuitos); em B, 5% das linhas teriam desconto maior que o próprio preço (p95 ~149%).

**Convenção usada daqui em diante:** receita líquida = `receita_aprovada`; receita bruta (preço cheio) = `receita_aprovada + vlr_venda_desconto`; taxa de desconto = `vlr_venda_desconto / (receita_aprovada + vlr_venda_desconto)`. É uma inferência a partir dos dados, a confirmar com o negócio.

### Números gerais do desconto

Antes das relações, o tamanho do desconto na base: quanto foi concedido, a taxa média e em quantas linhas, pedidos e itens ele aparece.

**Atenção ao grão:** a base é agregada por hora × canal × categoria, então não dá para saber se *cada pedido* teve desconto. O que medimos é a parcela dos pedidos/itens que está em **linhas com desconto > 0**.

### 2.3.4 Existe padrão por hora do dia?

Hora vai de **0 (meia-noite) a 23**. Para cada hora calculamos a **taxa** = linhas com o problema ÷ linhas da hora.

**Como saber se uma diferença é real ou acaso?** Se o problema não depende da hora, cada hora teria a mesma
taxa da base, com uma pequena oscilação aleatória. A **faixa cinza** é essa oscilação esperada (intervalo de
95%: taxa global ± 1,96 × √(p(1−p)/n)). Pontos **vermelhos** ficam fora da faixa.

**Cuidado com comparações múltiplas:** com 24 horas, mesmo sem padrão nenhum espera-se que ~1 hora
(5% de 24) caia fora da faixa por puro acaso. Por isso olhamos também o **teste qui-quadrado**, que avalia as
24 horas de uma vez: p-valor < 0,05 indica que a taxa **varia** com a hora. Um padrão operacional (ex.: carga
ou rotina noturna) apareceria como um **bloco de horas consecutivas** acima da faixa, não como pontos isolados.

```python
def taxa_por_hora(condicao_sql: str, titulo: str) -> pd.DataFrame:
    """Taxa de um problema por hora do dia, faixa de 95% esperada sem padrão e teste qui-quadrado."""
    sql = carregar_sql("21_sql_taxa_por_hora.sql", condicao_sql=condicao_sql)
    tabela = run_bq(sql, titulo)
    taxa_global = tabela["ocorrencias"].sum() / tabela["linhas"].sum()
    tabela["taxa"] = tabela["ocorrencias"] / tabela["linhas"]
    margem = 1.96 * np.sqrt(taxa_global * (1 - taxa_global) / tabela["linhas"])
    tabela["faixa_min"] = taxa_global - margem
    tabela["faixa_max"] = taxa_global + margem
    tabela["fora_da_faixa"] = (tabela["taxa"] < tabela["faixa_min"]) | (tabela["taxa"] > tabela["faixa_max"])
    tabela["lift_vs_global"] = tabela["taxa"] / taxa_global

    # qui-quadrado (tabela 24 horas x [com problema, sem problema])
    esperado = tabela["linhas"] * taxa_global
    qui2 = (((tabela["ocorrencias"] - esperado) ** 2 / esperado)
            + ((tabela["linhas"] - tabela["ocorrencias"] - (tabela["linhas"] - esperado)) ** 2
               / (tabela["linhas"] - esperado))).sum()
    graus = len(tabela) - 1
    try:
        from scipy.stats import chi2
        p_valor = chi2.sf(qui2, graus)
        texto_p = f"p-valor = {p_valor:.3f}"
    except ImportError:
        texto_p = f"(valor crítico a 5% com {graus} g.l. ≈ 35,2)"

    fig, ax = plt.subplots(figsize=(12, 4))
    ax.fill_between(tabela["hora"], tabela["faixa_min"] * 100, tabela["faixa_max"] * 100,
                    color="lightgray", alpha=0.6, label="faixa esperada sem padrão (95%)")
    ax.axhline(taxa_global * 100, color="gray", linestyle="--", label=f"taxa global {taxa_global:.2%}")
    ax.plot(tabela["hora"], tabela["taxa"] * 100, marker="o", label="taxa na hora")
    fora = tabela[tabela["fora_da_faixa"]]
    ax.scatter(fora["hora"], fora["taxa"] * 100, color="red", zorder=3, s=60, label="fora da faixa")
    ax.set_xticks(range(24))
    ax.set_title(f"{titulo} — qui-quadrado = {qui2:.1f} com {graus} g.l., {texto_p}")
    ax.set_xlabel("Hora do dia")
    ax.set_ylabel("% das linhas da hora")
    ax.legend(fontsize=8, loc="upper left", bbox_to_anchor=(1.01, 1))
    fig.tight_layout()
    plt.show()

    print(f"Horas fora da faixa: {int(tabela['fora_da_faixa'].sum())} de 24 "
          f"(esperado por acaso ≈ 1) -> {fora['hora'].tolist()}")
    return tabela


invalid_category_hour = taxa_por_hora(
    "SAFE_CAST(des_categoria_material AS NUMERIC) IS NOT NULL",
    "Categoria inválida por hora do dia",
)
display(invalid_category_hour[["hora", "linhas", "ocorrencias", "taxa", "lift_vs_global", "fora_da_faixa"]])
```

### 2.3.5 É possível recuperar as categorias inválidas?

O número gravado no campo é aleatório (2.3.3) e não diz nada sobre a categoria original. Mas o **grão da tabela** ajuda: existe **uma linha por hora × canal × categoria**, ou seja, no máximo 8 linhas por hora × canal (uma para cada categoria válida).

Se a linha inválida fosse uma venda *extra*, apareceriam horas com as 8 categorias válidas **e** uma inválida. Se ela for uma das 8 categorias que **perdeu o rótulo**, a soma de válidas + inválidas nunca passa de 8.

```python
sql_invalid_slot = carregar_sql("22_sql_invalid_slot.sql")

invalid_slot = run_bq(sql_invalid_slot, "categorias válidas × inválidas por hora × canal")

display(
    invalid_slot
    .pivot_table(
        index="categorias_validas",
        columns="categorias_invalidas",
        values="slots_hora_canal",
        fill_value=0,
    )
    .astype(int)
    .sort_index(ascending=False)
)

print(
    "Slots com mais de 8 linhas (válidas + inválidas):",
    int(invalid_slot.loc[invalid_slot["total_linhas_no_slot"] > 8, "slots_hora_canal"].sum()),
)
print(
    "Slots com 8 categorias válidas E alguma inválida:",
    int(invalid_slot.loc[
        (invalid_slot["categorias_validas"] == 8) & (invalid_slot["categorias_invalidas"] > 0),
        "slots_hora_canal",
    ].sum()),
)
```

**Leitura:** em nenhuma hora × canal a soma de válidas + inválidas passa de 8, e nunca existe uma linha inválida numa hora em que as 8 categorias já aparecem. Portanto a linha inválida **é sempre uma das 8 categorias que perdeu o nome**, não uma venda extra.

Isso permite a regra de recuperação: para cada linha inválida, listamos as categorias que **faltam** naquela hora × canal.

- **1 categoria faltante:** a categoria original só pode ser essa → **recuperável com segurança**.
- **2 ou mais faltantes:** a linha pode ser qualquer uma delas. É comum de madrugada, quando algumas categorias simplesmente não vendem.

```python
sql_invalid_recovery = carregar_sql("23_sql_invalid_recovery.sql")

invalid_recovery = run_bq(sql_invalid_recovery, "candidatas para cada categoria inválida")

invalid_recovery["situacao"] = pd.cut(
    invalid_recovery["n_candidatas"],
    bins=[0, 1, 2, np.inf],
    labels=["1 candidata (recuperável)", "2 candidatas", "3 ou mais candidatas"],
)

recovery_summary = (
    invalid_recovery
    .groupby("situacao", observed=True)
    .agg(linhas=("valor_invalido", "count"), qt_material=("qt_material", "sum"))
)
recovery_summary["pct_linhas"] = recovery_summary["linhas"] / recovery_summary["linhas"].sum()
recovery_summary["pct_qt_material"] = recovery_summary["qt_material"] / recovery_summary["qt_material"].sum()
display(recovery_summary)

print("Exemplos:")
display(
    invalid_recovery
    .sort_values("dt_hr_venda")
    [["dt_hr_venda", "des_canal_venda_final_agrup", "valor_invalido",
      "n_candidatas", "categorias_candidatas", "categoria_recuperada"]]
    .head(10)
)
```

#### E quando há 2 ou mais candidatas? O preço médio resolve?

A ideia natural seria escolher a candidata cujo **preço médio por item** (`receita_aprovada / qt_material`) fosse mais parecido com o da linha inválida. O problema: no grão hora × canal, o preço médio de uma mesma categoria **varia muito de uma hora para outra** (mix de produtos, promoções, kits), e as faixas das categorias se sobrepõem.

```python
sql_price_spread = carregar_sql("24_sql_price_spread.sql")

price_spread = run_bq(sql_price_spread, "variação do preço médio por hora, por categoria")
price_spread[["preco_medio_item", "p10", "p25", "mediana", "p75", "p90"]] = (
    price_spread[["preco_medio_item", "p10", "p25", "mediana", "p75", "p90"]].astype(float)
)
display(price_spread)

fig, ax = plt.subplots(figsize=(11, 5))
y = np.arange(len(price_spread))
ax.hlines(y, price_spread["p10"], price_spread["p90"], color="#9fb3c8", linewidth=6, label="p10–p90")
ax.hlines(y, price_spread["p25"], price_spread["p75"], color="#1f4e79", linewidth=10, label="p25–p75")
ax.scatter(price_spread["mediana"], y, color="white", edgecolor="black", zorder=3, label="mediana")
ax.set_yticks(y)
ax.set_yticklabels(price_spread["des_categoria_material"])
ax.set_xlabel("Preço médio por item na linha hora × canal (R$)")
ax.set_title("Preço médio por item varia muito entre as horas e as faixas se sobrepõem")
ax.grid(axis="x", alpha=0.3)
ax.legend(fontsize=8, loc="lower right")
fig.tight_layout()
plt.show()
```

As faixas p10–p90 das categorias se sobrepõem bastante (por exemplo FACIAL, CABELOS, MAQUIAGEM e CORPO E BANHO ocupam praticamente a mesma região). Para medir o quanto o preço é confiável como "classificador", testamos nas **linhas válidas**, onde sabemos a resposta certa:

- **Entre as 8 categorias:** a categoria de preço de referência mais próximo é a verdadeira?
- **Entre 2 categorias** (o caso ambíguo mais comum): comparando a categoria verdadeira com cada uma das outras 7, com que frequência o preço escolhe a certa?

O preço de referência é o preço médio ponderado da categoria no mesmo mês e canal.

```python
sql_price_classifier = carregar_sql("25_sql_price_classifier.sql")

price_classifier = run_bq(sql_price_classifier, "preço médio como classificador de categoria")
display(price_classifier)
```

**Conclusão — é possível recuperar?**

- **Sim, quando há 1 candidata: 3.072 linhas (57,5% das inválidas e 62,5% do `qt_material` inválido).** A estrutura do grão garante que a linha inválida é a única categoria ausente naquela hora × canal, então a recuperação é **determinística**, sem estimativa. Com isso, a categoria inválida cai de ~6,0% para ~2,5% das linhas.
- **Não, quando há 2 ou mais candidatas: 2.269 linhas (42,5%).** O preço médio por item **não é confiável** para desempatar:
  - o preço de uma mesma categoria varia muito de hora para hora (FACIAL vai de R$ 15 a R$ 94 entre p10 e p90; PERFUMARIA FEMININA de R$ 38 a R$ 139) e as faixas se sobrepõem;
  - mesmo nas linhas **válidas**, onde a resposta é conhecida, o preço acerta a categoria em só **33,5%** dos casos entre 8 opções e, entre apenas 2 opções, erra **~1 em cada 4** (77,2% de acerto, contra 50% por acaso).

**Decisão:** recuperar apenas os casos de 1 candidata e manter os demais como `CATEGORIA_INVALIDA`. Os totais diários (target) não mudam, pois essas linhas já estavam incluídas; a recuperação melhora somente as análises e features por categoria.

#### Diagnóstico — categoria inválida

- **5,96% das linhas** têm categoria numérica, com taxa constante no tempo, nos dois canais e em todas as horas: falha estrutural de integração, não comportamento de venda. As vendas dessas linhas são legítimas.
- **Recuperar só o que é determinístico:** quando falta exatamente **1 categoria** naquela hora × canal (57,5% das linhas inválidas), ela é a categoria original → corrigir na camada tratada.
- **Não recuperar o resto:** com 2 ou mais candidatas não há assertividade (o preço médio por item erra ~1 em 4 mesmo escolhendo entre 2) → manter como `CATEGORIA_INVALIDA`.
- **Totais não mudam:** as linhas continuam no total diário (target); a recuperação só melhora as análises por categoria.

**O que fazer:** **warning a cada ocorrência** — toda carga que trouxer qualquer linha com categoria numérica gera um aviso (check `categoria_invalida`, seção 2.8), com a hora, o canal e o valor recebido, para que o time de origem entenda por que isso acontece. A carga não é bloqueada: a linha segue para os totais e, quando possível, a categoria é recuperada pela regra de 1 candidata. O objetivo é zerar a ocorrência corrigindo a causa na origem.

### 2.4.7 É erro de sinal? E o desconto explica alguma coisa?

Três explicações possíveis para a receita negativa, cada uma com uma "pegada" diferente nos dados:

1. **Devolução/estorno:** itens, pedidos e desconto também viriam negativos.
2. **Prejuízo estratégico** (desconto maior que o preço): o preço cheio (`receita_aprovada + vlr_venda_desconto`) ficaria negativo ou próximo de zero.
3. **Erro de sinal:** com o sinal corrigido (`ABS(receita_aprovada)`), a linha teria o mesmo perfil de uma venda normal da mesma categoria — preço por item, desconto por item e taxa de desconto.

A taxa de desconto calculada com a receita negativa não tem significado (o denominador fica pequeno ou negativo, gerando taxas acima de 100%). Por isso comparamos com o sinal corrigido.

```python
sql_negative_sign = carregar_sql("32_sql_negative_sign.sql")
display(run_bq(sql_negative_sign, "receita negativa: sinais das outras colunas").T.rename(columns={0: "valor"}))

sql_negative_profile = carregar_sql("33_sql_negative_profile.sql")
negative_profile = run_bq(sql_negative_profile, "receita negativa: perfil com sinal corrigido")
display(
    negative_profile
    .pivot(index="des_categoria_material", columns="grupo",
           values=["preco_medio_item", "desconto_por_item", "taxa_desconto"])
    .round(2)
)

sql_negative_discount_dist = carregar_sql("34_sql_negative_discount_dist.sql")
display(run_bq(sql_negative_discount_dist, "receita negativa: distribuição da taxa de desconto"))
```

**Leitura:**

- **Não é devolução:** todas as linhas negativas têm `qt_material` e `nr_pedidos` positivos e nenhuma tem desconto negativo — só a receita veio negativa.
- **Não é prejuízo estratégico:** nessa leitura o preço cheio ficaria **negativo ou zero** na grande maioria das linhas, o que é impossível. Brindes aparecem na base como receita **zero**, não negativa (ver 1.4).
- **É compatível com erro de sinal:** com `ABS(receita_aprovada)`, preço médio por item, desconto por item e taxa de desconto ficam praticamente **iguais** aos das linhas positivas da mesma categoria, e a distribuição da taxa de desconto é a mesma (mediana ~31% nos dois grupos).
- **O desconto não explica a receita negativa:** ele tem o mesmo perfil nas linhas negativas e positivas. A aparente concentração de receita negativa nas faixas de "maior desconto" vista antes era um artefato de calcular a taxa com a receita negativa no denominador.

#### Diagnóstico — receita negativa

- **3,92% das linhas** (R$ 21 mi em valor absoluto) têm receita negativa, distribuídas de forma uniforme: sem padrão por canal, categoria, mês ou hora, e sem relação com o desconto.
- Itens, pedidos e desconto são positivos, e com o sinal corrigido a linha tem exatamente o perfil de uma venda normal → **erro de sinal** (inferência forte, a confirmar com o negócio).

**O que fazer:**

- **Corrigir na camada tratada:** `ABS(receita_aprovada)` + uma coluna `flag_receita_sinal_corrigido` para rastreabilidade; a `stg_fact_vendas` preserva o valor original. Com a correção, a receita do período sobe ~8,6% (de R$ 490 mi para ~R$ 532 mi).
- **Warning por hora, a cada ocorrência** (check `receita_negativa`, seção 2.8): toda carga com receita negativa gera aviso para o time de origem, como na categoria inválida — é um erro a ser eliminado, não uma taxa a ser tolerada.
- Levar ao time de engenharia de dados junto com a categoria inválida: os dois erros têm o mesmo comportamento aleatório e podem ter a mesma origem.

```python
sql_daily_outliers = carregar_sql("40_sql_daily_outliers.sql")

outlier_limits = run_bq(sql_daily_outliers, "limites de outliers")
display(outlier_limits)

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

axes[0].boxplot(daily["qt_material"], vert=False)
axes[0].set_title("Boxplot — itens diários (target)")
axes[0].set_xlabel("qt_material")

axes[1].boxplot(daily["nr_pedidos"], vert=False)
axes[1].set_title("Boxplot — pedidos diários")
axes[1].set_xlabel("nr_pedidos")

axes[2].boxplot(daily["receita_aprovada"], vert=False)
axes[2].set_title("Boxplot — receita diária")
axes[2].set_xlabel("receita_aprovada")

fig.tight_layout()
plt.show()

# Diagnóstico complementar: os mesmos picos aparecem em qt_material e nr_pedidos?
daily_outlier_flags = daily[[
    "data",
    "qt_material",
    "nr_pedidos",
    "receita_aprovada",
    "itens_por_pedido",
]].copy()

limits = outlier_limits.iloc[0]

daily_outlier_flags["outlier_qt_material"] = (
    (daily_outlier_flags["qt_material"] < limits["li_qt_material"])
    | (daily_outlier_flags["qt_material"] > limits["ls_qt_material"])
)
daily_outlier_flags["outlier_nr_pedidos"] = (
    (daily_outlier_flags["nr_pedidos"] < limits["li_nr_pedidos"])
    | (daily_outlier_flags["nr_pedidos"] > limits["ls_nr_pedidos"])
)
daily_outlier_flags["outlier_receita_aprovada"] = (
    (daily_outlier_flags["receita_aprovada"] < limits["li_receita_aprovada"])
    | (daily_outlier_flags["receita_aprovada"] > limits["ls_receita_aprovada"])
)

outlier_overlap = pd.DataFrame({
    "metrica": [
        "Outlier de qt_material",
        "Outlier de nr_pedidos",
        "Outlier de receita_aprovada",
        "qt_material + nr_pedidos",
        "qt_material sem nr_pedidos",
        "nr_pedidos sem qt_material",
        "qt_material + nr_pedidos + receita_aprovada",
    ],
    "dias": [
        int(daily_outlier_flags["outlier_qt_material"].sum()),
        int(daily_outlier_flags["outlier_nr_pedidos"].sum()),
        int(daily_outlier_flags["outlier_receita_aprovada"].sum()),
        int((daily_outlier_flags["outlier_qt_material"] & daily_outlier_flags["outlier_nr_pedidos"]).sum()),
        int((daily_outlier_flags["outlier_qt_material"] & ~daily_outlier_flags["outlier_nr_pedidos"]).sum()),
        int((~daily_outlier_flags["outlier_qt_material"] & daily_outlier_flags["outlier_nr_pedidos"]).sum()),
        int((
            daily_outlier_flags["outlier_qt_material"]
            & daily_outlier_flags["outlier_nr_pedidos"]
            & daily_outlier_flags["outlier_receita_aprovada"]
        ).sum()),
    ],
})

display(outlier_overlap)

display(
    daily_outlier_flags[
        daily_outlier_flags[
            ["outlier_qt_material", "outlier_nr_pedidos", "outlier_receita_aprovada"]
        ].any(axis=1)
    ].sort_values("qt_material", ascending=False)
)
```

### Como interpretar estes boxplots

- **Caixa:** vai de Q1 (25% dos dias abaixo) até Q3 (75% dos dias abaixo). Portanto contém os 50% centrais da distribuição.
- **Linha laranja dentro da caixa:** mediana, isto é, metade dos dias abaixo e metade acima.
- **Bigodes:** chegam até a observação mais extrema que ainda esteja dentro dos limites de Tukey.
- **Círculos fora dos bigodes:** dias sinalizados como outliers estatísticos. Cada círculo é um dia.

Como o limite inferior é negativo (≈ -5.081), ele não é relevante para uma quantidade vendida não negativa. Assim, os **26 dias** sinalizados são todos dias acima de ~68,5 mil itens. Em `nr_pedidos` são **25 dias** acima de ~47,6 mil pedidos.

Para receita: Q1 ≈ **R$ 1,35 mi**, Q3 ≈ **R$ 2,25 mi** e limite superior ≈ **R$ 3,59 mi**. Existem **25 dias** sinalizados. A grande concentração de pontos à direita indica uma distribuição assimétrica, típica de e-commerce com campanhas e datas comerciais.

### Por que olhar pedidos também?

Um dia extremo em `qt_material` pode ter duas explicações diferentes: **mais pedidos** ou **mais itens por pedido**. Ao cruzar outliers de `qt_material` com `nr_pedidos` e observar `itens_por_pedido`, conseguimos distinguir um pico de tráfego/conversão de uma mudança no tamanho da cesta. Dos 26 dias extremos em `qt_material`, 23 também são extremos em `nr_pedidos`.


**Por que o IQR do período não serve como alerta:** dos 26 dias acima do limite em `qt_material`, praticamente todos são de **novembro** (Black Friday e o mês de campanha que a antecede). O IQR compara cada dia com o período inteiro, então marca o mês todo como "anormal" e não diz nada sobre os outros meses. Ele é útil como retrato, não como regra de monitoramento.

Também testamos o IQR por categoria × canal: sobravam centenas de extremos, também concentrados em novembro — ruído demais para virar alerta.

### Alerta de outlier: comparar cada dia com o seu **valor esperado**

Em vez de comparar com o período inteiro, cada dia é comparado com o que seria esperado **para ele**, usando só informação do passado. O alerta cobre duas métricas: **volume** (`qt_material`) e **desconto** (taxa de desconto do dia).

1. **Esperado:**
   - volume: mediana dos 28 dias anteriores, ajustada pelo dia da semana (fator calculado fora de novembro);
   - desconto: mediana da taxa dos 28 dias anteriores.
2. **Desvio:** volume em `log(qt_material / esperado)`; desconto em pontos percentuais (`taxa - esperada`).
3. **Tolerância:** desvio robusto (MAD) medido **uma vez** num período estável (15/01 a 30/04, sem Black Friday, Natal ou Dia das Mães) e mantido fixo. O alerta dispara quando o desvio passa de **3 vezes** a tolerância.

Com a tolerância calibrada uma vez, a avaliação começa já na terceira semana de novembro (só o nível recente precisa de 14 dias de histórico), e a Black Friday entra na avaliação.

```python
LIMITE_ALERTA = 3.0
PERIODO_CALIBRACAO = ("2026-01-15", "2026-04-30")  # período estável, sem grandes eventos

alerta = daily[["data", "qt_material", "receita_aprovada", "vlr_venda_desconto"]].copy().set_index("data")
alerta["qt_material"] = alerta["qt_material"].astype(float)
alerta["taxa_desconto_pp"] = 100 * (
    alerta["vlr_venda_desconto"] / (alerta["receita_aprovada"] + alerta["vlr_venda_desconto"])
).astype(float)
calibracao = (alerta.index >= PERIODO_CALIBRACAO[0]) & (alerta.index <= PERIODO_CALIBRACAO[1])


def mad(serie: pd.Series) -> float:
    """Desvio absoluto mediano, na escala do desvio-padrão."""
    return 1.4826 * np.median(np.abs(serie - np.median(serie)))


# Volume: nível recente sem efeito do dia da semana, só com dias anteriores (shift)
fora_novembro = alerta[alerta.index.month != 11]
fator_dia_semana = (
    fora_novembro.groupby(fora_novembro.index.dayofweek)["qt_material"].median()
    / fora_novembro["qt_material"].median()
)
alerta["fator_dia_semana"] = alerta.index.dayofweek.map(fator_dia_semana)
alerta["qt_material_esperado"] = (
    (alerta["qt_material"] / alerta["fator_dia_semana"]).shift(1).rolling(28, min_periods=14).median()
    * alerta["fator_dia_semana"]
)
alerta["desvio_volume"] = np.log(alerta["qt_material"] / alerta["qt_material_esperado"])
tolerancia_volume = mad(alerta.loc[calibracao, "desvio_volume"].dropna())

# Desconto: taxa do dia contra a mediana dos 28 dias anteriores, em p.p.
alerta["taxa_desconto_esperada_pp"] = alerta["taxa_desconto_pp"].shift(1).rolling(28, min_periods=14).median()
alerta["desvio_desconto_pp"] = alerta["taxa_desconto_pp"] - alerta["taxa_desconto_esperada_pp"]
tolerancia_desconto_pp = mad(alerta.loc[calibracao, "desvio_desconto_pp"].dropna())
LIMITE_DESCONTO_PP = round(LIMITE_ALERTA * tolerancia_desconto_pp, 1)  # reutilizado nos checks da 2.8

alerta["score_volume"] = alerta["desvio_volume"] / tolerancia_volume
alerta["score_desconto"] = alerta["desvio_desconto_pp"] / tolerancia_desconto_pp
for metrica in ["volume", "desconto"]:
    alerta[f"alerta_{metrica}"] = np.select(
        [alerta[f"score_{metrica}"] > LIMITE_ALERTA, alerta[f"score_{metrica}"] < -LIMITE_ALERTA],
        ["ALTA", "QUEDA"], default="",
    )

print(f"Tolerância do volume: {tolerancia_volume:.3f} (log) | tolerância do desconto: {tolerancia_desconto_pp:.1f} p.p. "
      f"→ alerta de desconto acima de ±{LIMITE_DESCONTO_PP} p.p. do esperado")
print(f"Dias avaliados: {alerta['score_volume'].notna().sum()} (a partir de {alerta['score_volume'].first_valid_index():%d/%m/%Y})")

dias_alerta = alerta[(alerta["alerta_volume"] != "") | (alerta["alerta_desconto"] != "")].copy()
dias_alerta["variacao_volume"] = dias_alerta["qt_material"] / dias_alerta["qt_material_esperado"] - 1
display(dias_alerta[[
    "qt_material", "qt_material_esperado", "variacao_volume", "alerta_volume",
    "taxa_desconto_pp", "taxa_desconto_esperada_pp", "alerta_desconto",
]].round(2))

fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)
paineis = [
    (axes[0], "qt_material", "qt_material_esperado", "alerta_volume", "qt_material diário"),
    (axes[1], "taxa_desconto_pp", "taxa_desconto_esperada_pp", "alerta_desconto", "taxa de desconto do dia (%)"),
]
for ax, valor, esperado, coluna_alerta, titulo in paineis:
    ax.plot(alerta.index, alerta[valor], color="#1f4e79", alpha=0.7, label=titulo)
    ax.plot(alerta.index, alerta[esperado], color="grey", linestyle="--", linewidth=1, label="esperado")
    for tipo, cor in [("ALTA", "#6a994e"), ("QUEDA", "firebrick")]:
        pontos = alerta[alerta[coluna_alerta] == tipo]
        ax.scatter(pontos.index, pontos[valor], color=cor, s=50, zorder=3, label=f"alerta de {tipo.lower()}")
    ax.set_title(f"Alerta de outlier — {titulo} × valor esperado")
    ax.legend(fontsize=8, loc="upper center")
    ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
```

**Leitura:**

- **Volume — 5 alertas**, todos explicados pelo calendário: **Black Friday** (28/11, +262% sobre o esperado), a **queda logo depois** (02 e 03/12, ~−70%: o "esperado" ainda carrega o patamar da campanha de novembro), **Natal** (25/12) e **Ano-Novo** (01/01), dias de venda quase parada.
- **Desconto — 5 alertas:** a **queda do desconto no fim da Black Friday** (02 e 03/12: de ~55% para ~31%) e três dias de **desconto muito acima do normal: 30/05, 29/06 e 30/06** (~55% contra ~29–33% esperado).
- **O desconto explica o pico do fim de junho:** 29 e 30/06 eram os dias de volume alto sem evento no calendário (+125% e +183% sobre o esperado). O alerta de desconto mostra que foram dias de **promoção forte**, com o desconto quase dobrando — não um erro de dado. O mesmo vale para 30/05: parece haver uma **campanha de fim de mês**, a confirmar com o negócio.
- **Por que maio (Dia das Mães) não alerta mais:** com a tolerância calibrada no período estável, a alta da semana do Dia das Mães (+130% a +160% sobre o esperado) fica abaixo do limite de 3. É um pico relevante, mas dentro do que o alerta considera variação de campanha; o check horário da 2.8 ainda o captura (09/05).
- No gráfico de desconto também aparece o ritmo das **campanhas: ondas de 1 a 2 semanas** em que a taxa sobe ~15 p.p. e depois volta.

Por ser o grande evento do varejo de perfumaria, a Black Friday é tratada como uma **campanha**: a janela de comparação cobre novembro inteiro e a semana seguinte.
- Fora de novembro, os maiores dias coincidem com o calendário: **Dia do Consumidor** (15/03, exatamente no dia) e a semana do **Dia das Mães** (08/05). O Natal aparece como **queda**, não como pico.
- **Dia da Mulher (08/03)** não forma um pico próprio: a alta de março começa em 03/03 e só termina depois de 15/03, quando o volume cai pela metade de um dia para o outro. O formato acompanha a semana do Dia do Consumidor (desconto de ~40–44% no período), e não a data de 08/03.
- **29–30/06** (o maior volume fora de novembro, sem evento no calendário) é explicado pelo **desconto**: a taxa quase dobrou (~55% contra ~28%), o que indica promoção de fim de mês.
- **Não remover outliers:** são vendas reais. Para o forecast, a informação certa é o **calendário comercial** e o **plano de desconto** como variáveis explicativas, não a exclusão desses dias.

**O que fazer:**

- **Alerta no nível do total** (não por categoria × canal) para **volume e desconto**: a cada carga horária, o check `volume_anomalo_3h` (seção 2.8); no fechamento do dia, o alerta diário acima e o check `taxa_desconto_anomala` (desconto fora de ±22 p.p. do esperado).
- **Triagem de cada alerta:** (1) o dia está completo (24 horas, número normal de linhas)? Se não, é **pipeline**; (2) há evento no calendário? Se sim, **esperado**; (3) houve mudança forte de desconto? Se sim, **campanha** — confirmar com o negócio; (4) se nada disso, investigar.
- **Por categoria** apenas como monitoramento secundário, com limite mais alto e só para as categorias grandes: no nível categoria × canal a série é ruidosa demais e geraria alarmes falsos.

## 2.8 Automação: checks a cada carga horária

As seções 2.1 a 2.7 investigaram o histórico. Em produção, os dados chegam **hora a hora**, e cada carga precisa ser verificada automaticamente antes de alimentar análises e o forecast. Aqui as regras viram **checks** (no Dataform seriam *assertions*: consultas que retornam o que viola a regra), em dois níveis de severidade:

- **ERRO** — bloqueia a carga: o dado não pode seguir para análises nem para o modelo.
- **WARNING** — não bloqueia, mas notifica para triagem.

| Check | Frequência | Severidade | Regra |
|---|---|---|---|
| `nulos` | a cada hora | ERRO | nenhuma coluna nula |
| `dominio` | a cada hora | ERRO | canal ∈ {App, Site}; `qt_material` e `nr_pedidos` > 0 |
| `duplicidade_grao` | a cada hora | ERRO | uma linha por hora × canal × categoria |
| `hora_sem_dados` | a cada hora | WARNING | a hora tem ao menos uma linha |
| `canal_ausente` | a cada hora | WARNING | os dois canais presentes na hora |
| `categoria_invalida` | a cada hora | WARNING | **qualquer** linha com categoria numérica gera aviso (toda ocorrência é investigada) |
| `receita_negativa` | a cada hora | WARNING | **qualquer** linha com receita negativa gera aviso (erro de sinal a ser corrigido na origem, 2.4) |
| `volume_anomalo_3h` | a cada hora | WARNING | `qt_material` entre 0,3× e 3× o esperado (mediana da mesma hora e dia da semana nas 4 semanas anteriores), disparando só se 3 horas seguidas ficarem fora |
| `taxa_desconto_anomala` | fechamento do dia | WARNING | taxa de desconto do dia a até ±22 p.p. da mediana dos 28 dias anteriores (limite calibrado na 2.7) |

Receita negativa e categoria inválida são **erros que o time de origem precisa eliminar**, então não usam limite de taxa: **qualquer ocorrência**, em qualquer hora, gera aviso. A meta é que esses checks cheguem a zero.

Abaixo, a mesma consulta é aplicada a **todo o histórico**, como se cada hora tivesse chegado em uma carga — um *backtest* dos checks, para calibrar quantos alertas eles gerariam.

```python
sql_dq_checks = carregar_sql("41_sql_dq_checks.sql", LIMITE_DESCONTO_PP=LIMITE_DESCONTO_PP)

dq_checks = run_bq(sql_dq_checks, "checks de Data Quality por carga (backtest no histórico)")
dq_checks["referencia"] = pd.to_datetime(dq_checks["referencia"])

print("Resumo do backtest: quantas vezes cada check teria disparado")
display(
    dq_checks
    .groupby(["frequencia", "check_nome", "severidade"], as_index=False)
    .agg(avaliacoes=("falhou", "size"), disparos=("falhou", "sum"))
)

# Warnings agrupados por dia (horas consecutivas viram um único incidente) e triagem pelo calendário
disparos = dq_checks[dq_checks["falhou"]].copy()
disparos["dia"] = disparos["referencia"].dt.normalize()
incidentes = (
    disparos
    .groupby(["dia", "check_nome", "severidade"], as_index=False)
    .agg(horas=("referencia", "size"), exemplo=("detalhe", "first"))
)
incidentes["evento_proximo"] = incidentes["dia"].apply(evento_proximo)

def triagem(linha) -> str:
    """Primeira leitura automática de cada incidente."""
    if linha["check_nome"] == "volume_anomalo_3h":
        return "evento comercial: esperado" if linha["evento_proximo"] else "investigar (pipeline ou negócio)"
    if linha["check_nome"] in ("hora_sem_dados", "canal_ausente"):
        return "conferir volume típico da hora (assertion da 2.5)"
    if linha["check_nome"] == "categoria_invalida":
        return "recuperar se houver 1 candidata (2.3.5) e reportar à origem"
    if linha["check_nome"] == "receita_negativa":
        return "corrigir sinal com flag (2.4) e reportar à origem"
    if linha["check_nome"] == "taxa_desconto_anomala":
        return "evento comercial: esperado" if linha["evento_proximo"] else "confirmar campanha com o negócio"
    return "investigar com o time de origem"

incidentes["triagem"] = incidentes.apply(triagem, axis=1)
# categoria_invalida e receita_negativa disparam em quase todos os dias (toda ocorrência é aviso): resumo separado
por_ocorrencia = ["categoria_invalida", "receita_negativa"]
total_dias = dq_checks["referencia"].dt.normalize().nunique()
for check in por_ocorrencia:
    avisos = incidentes[incidentes["check_nome"] == check]
    print(f"{check}: aviso em {len(avisos)} de {total_dias} dias, {int(avisos['horas'].sum())} horas no total")
outros_incidentes = incidentes[~incidentes["check_nome"].isin(por_ocorrencia)]
print(f"Demais incidentes: {len(outros_incidentes)}")
display(outros_incidentes)

ultima_hora = dq_checks.loc[dq_checks["frequencia"] == "hora", "referencia"].max()
print(f"Status da última carga horária ({ultima_hora:%d/%m/%Y %H:%M}):")
display(
    dq_checks[(dq_checks["frequencia"] == "hora") & (dq_checks["referencia"] == ultima_hora)]
    [["check_nome", "severidade", "falhou", "detalhe"]]
)
```

#### Diagnóstico — automação

- **Checks de ERRO:** nenhum disparo em 5.808 horas. A base histórica é estruturalmente íntegra, e essas regras ficam como proteção para as próximas cargas.
- **`categoria_invalida` e `receita_negativa`:** disparam em **todos os 242 dias** (3.573 e 2.658 horas) — esperado, já que os dois erros são estruturais e ainda não foram corrigidos na origem. Como **toda ocorrência é aviso**, o volume é alto: para não virar ruído, os avisos podem ser consolidados num **resumo diário** (horas, linhas e valores) enviado ao time de origem, até a causa ser corrigida e os checks voltarem a zero.

### Por que não modelar diretamente no grão horário?

O arquivo horário é valioso e será explorado, mas o **target permanece diário** porque o case solicita previsão de venda diária.

O grão horário pode ser usado a nosso favor para:
- entender o perfil intradiário;
- criar, se necessário, features **históricas** de comportamento horário;
- criar lags por canal/categoria/hora.

Não devemos usar informações do próprio dia realizado para prever esse mesmo dia. Modelar hora a hora e depois somar seria possível, mas adicionaria complexidade sem responder melhor ao objetivo principal do case. A estratégia inicial, portanto, é **forecast diário + enriquecimento com sinais históricos da granularidade horária**.

### Baseline imediato: médias móveis de 7 e 28 dias

As médias móveis do gráfico acima são descritivas. Para tratá-las como **previsões**, precisamos deslocá-las um dia para trás, garantindo que a previsão de *t* use apenas informação disponível até *t-1*.

Usamos uma janela comum após 28 dias para comparar os dois baselines de forma justa. A métrica principal é **WAPE**. Também mostramos `1 - WAPE` como uma leitura percentual intuitiva, mas ela não substitui as métricas formais do modelo.

```python
baseline = daily[["data", "qt_material"]].copy()

baseline["pred_media_movel_7"] = (
    baseline["qt_material"]
    .shift(1)
    .rolling(7, min_periods=7)
    .mean()
)

baseline["pred_media_movel_28"] = (
    baseline["qt_material"]
    .shift(1)
    .rolling(28, min_periods=28)
    .mean()
)

evaluation = baseline.dropna(
    subset=["pred_media_movel_7", "pred_media_movel_28"]
).copy()

def forecast_metrics(y_true, y_pred):
    error = y_pred - y_true
    abs_error = np.abs(error)

    mae = abs_error.mean()
    wape = abs_error.sum() / np.abs(y_true).sum()
    bias = error.sum() / y_true.sum()

    return {
        "MAE": mae,
        "WAPE": wape,
        "1-WAPE": 1 - wape,
        "Bias": bias,
    }

baseline_metrics = pd.DataFrame([
    {
        "modelo": "Média móvel 7 dias",
        **forecast_metrics(
            evaluation["qt_material"],
            evaluation["pred_media_movel_7"],
        ),
    },
    {
        "modelo": "Média móvel 28 dias",
        **forecast_metrics(
            evaluation["qt_material"],
            evaluation["pred_media_movel_28"],
        ),
    },
])

display(baseline_metrics)

fig, ax = plt.subplots(figsize=(13, 5))
ax.plot(evaluation["data"], evaluation["qt_material"], label="Real", alpha=0.65)
ax.plot(evaluation["data"], evaluation["pred_media_movel_7"], label="MM 7 dias")
ax.plot(evaluation["data"], evaluation["pred_media_movel_28"], label="MM 28 dias")
ax.set_title("Backtest simples — médias móveis como baseline")
ax.set_xlabel("Data")
ax.set_ylabel("qt_material")
ax.legend()
fig.tight_layout()
plt.show()
```

# 5. Model Readiness

### Target proposto
**`qt_material` diário**: demanda em unidades, que é o que a operação planeja (estoque, reposição, separação). Também é aditivo entre categorias e canais.

`nr_pedidos` fica como métrica de apoio. Como vem agregado por hora × categoria × canal, um pedido com itens de duas categorias provavelmente é contado nas duas; a soma diária não é o número de pedidos distintos (hipótese a validar com o negócio). `receita_aprovada` permanece como KPI financeiro.


### Risco de leakage
Para prever o dia *t*, métricas realizadas do próprio dia *t* não podem ser features: receita, pedidos, itens, desconto e mix realizado do mesmo período só existem depois da venda.

### Features candidatas
- calendário: dia da semana e dia do mês;
- datas comerciais (calendário de eventos conhecido antes da previsão) e distância até/desde cada data;
- plano de desconto: taxa da semana por canal × categoria, premissa informada pelo time de desconto;
- canal e categoria (dummies), que definem o nível de cada série;
- lags e médias móveis **só com defasagem ≥ 14 dias** (o horizonte de previsão): com previsão de 14 dias à frente, `t-1` e `t-7` ainda não existem no momento da previsão.

### Validação
Split aleatório não é apropriado. Usamos **backtest temporal / expanding window**: cada rodada treina com tudo até a véspera e prevê os 14 dias seguintes.

### Baselines obrigatórios
- naive sazonal (repete a última semana observada);
- médias móveis de 7 e 28 dias (a de 7 dias é a referência na apresentação).

### Métricas
WAPE, MAE, viés e skill vs baseline. MAPE será tratado com cautela.

O resultado está no notebook `modelo_machine_learning_categorias.ipynb`: os lags foram testados e não melhoraram o modelo, então ficaram de fora; o modelo final é o LightGBM com calendário, datas comerciais e plano de desconto.
